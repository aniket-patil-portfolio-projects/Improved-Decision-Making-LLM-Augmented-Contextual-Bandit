```python
# import the required libraries
import pandas as pd
from transformers import BertTokenizer, BertModel
import torch
import numpy as np

# read the input data (embedded keywords dataset) into a pandas dataframe
""" 
the embedded keywords dataset is the dataset 
formed by merging the embedded chunks 1-4 generated by the files named BERT_Embeddings_Keywords_X. This dataset was merged manually.
"""
movies_df = pd.read_csv("/content/drive/MyDrive/LLM-NEW-2/bert-embed/embedded-data/mkg_keywords_dataset_bert.csv")

# load the pre-trained bert tokenizer and model
model_name='bert-base-uncased'
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertModel.from_pretrained(model_name)

"""
:function: create_word_embeddings
:param sentence: takes a input sentence to be embedded i.e., movie title
:return: embeded sentence vector of size 768
""" 
def create_word_embeddings(sentence):
    input_ids = tokenizer.encode(sentence, return_tensors='pt')
    with torch.no_grad():
        output = model(input_ids)
    embeddings = output.last_hidden_state.mean(dim=1).squeeze().numpy()
    return embeddings

# Apply the function to each row and create a new column 'keywords_embedding'
movies_df['original_title_embedding'] = movies_df['original_title'].apply(create_word_embeddings)

# export the embedded title data to a .csv file
movies_df.to_csv('/content/drive/MyDrive/LLM-NEW-2/gpt2-embed/embedded-data/mkg_title_dataset_bert.csv', index=False)
```
