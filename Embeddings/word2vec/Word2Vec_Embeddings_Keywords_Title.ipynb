```python
# import the word2vec model
from gensim.models import Word2Vec
```

```python
# import the required libraries
import pandas as pd
import torch
import numpy as np

# read the input data (embedded keywords-title gpt2 dataset) into a pandas dataframe.
""" 
the embedded keywords-title gpt2 dataset is the dataset 
formed by merging the embedded keywords and embedded title dataset for gpt2
"""
movies_df = pd.read_csv("/content/drive/MyDrive/LLM-NEW-2/gpt2-embed/input/final_mkg_dataset_gpt2.csv")

movies_df.shape[0]
```

```python
movies_df.drop(columns=['keywords_embedding_string', 'original_title_embedding_string'], inplace=True)
```

```python
movies_df.head(10)
```

```python
"""
:function: create_embeddings
:params: movies_df
:return: embedded vector

This function is applied to each row of the movies_df dataset to get the word2vec embeddings 
for the keyword column.
The model is first loaded, with various model parameters and trained on the input keywords
followed by embedding creation. The size of the embedded vector is 768.
"""
def create_embeddings(movies_df):
  all_keywords = movies_df['keywords'].sum().split()
  model = Word2Vec(sentences=[all_keywords], vector_size=768, min_count=1, workers=3, iterations=5, window=5)
  word_embeddings = {}

  def get_embedding(keywords):
    keyword_list = keywords.split()
    embeddings = []
    for word in keyword_list:
      if word in word_embeddings:
        embeddings.append(word_embeddings[word])
      else:
        try:
          embedding = model.wv[word]
          word_embeddings[word] = embedding
          embeddings.append(embedding)
        except KeyError:
          pass
    if embeddings:
      return np.mean(embeddings, axis=0)
    else:
      return np.zeros(model.vector_size)

  movies_df['keywords_embedding'] = movies_df['keywords'].apply(get_embedding)

create_embeddings(movies_df)
```

```python
"""
:function: create_embeddings
:params: movies_df
:return: embedded vector

This function is applied to each row of the movies_df dataset to get the word2vec embeddings 
for the keyword column.
The model is first loaded, with various model parameters and trained on the input title
followed by embedding creation. The size of the embedded vector is 768.
"""
def create_embeddings(movies_df):
  all_title = movies_df['original_title'].sum().split()
  model = Word2Vec(sentences=[all_title], vector_size=768, min_count=1)
  word_embeddings = {}

  def get_embedding(titles):
    title_list = titles.split()
    embeddings = []
    for word in title_list:
      if word in word_embeddings:
        embeddings.append(word_embeddings[word])
      else:
        try:
          embedding = model.wv[word]
          word_embeddings[word] = embedding
          embeddings.append(embedding)
        except KeyError:
          pass
    if embeddings:
      return np.mean(embeddings, axis=0)
    else:
      return np.zeros(model.vector_size)

  movies_df['original_title_embedding'] = movies_df['original_title'].apply(get_embedding)

create_embeddings(movies_df)
```
