```python
# import the word2vec model
from gensim.models import Word2Vec
```

```python
# import the required libraries
import pandas as pd
import torch
import numpy as np

# read the input data (embedded keywords-title word2vec dataset) into a pandas dataframe.
""" 
the embedded keywords-title word2vec dataset is the dataset 
formed by merging the embedded keywords and embedded title dataset for word2vec
"""
movies_df = pd.read_csv("/content/drive/MyDrive/LLM-NEW-2/word2vec-embed/input/old_final_mkg_dataset_word2vec.csv")
```

```python
drop_columns = ['adult', 'genres', 'original_language',
       'popularity', 'release_date',
       'spoken_languages', 'vote_average', 'vote_count', 'movie_id',
       'keywords', 'genres_indices', 'keywords_embedding_string',
       'original_title_embedding_string']
movies_df = movies_df.drop(columns=drop_columns)
movies_df.head(5)
```

```python
# reduce the size of the overview column to 500 characters of string
max_length = movies_df['overview'].str.len().max()
print("Maximum length of column '{}': {}".format('overview', max_length))
movies_df['overview'] = movies_df['overview'].str.slice(0, 500)
max_length = movies_df['overview'].str.len().max()
print("Maximum length of column after strip '{}': {}".format('overview', max_length))
```

```python
"""
:function: create_embeddings
:params: movies_df
:return: embedded vector

This function is applied to each row of the movies_df dataset to get the word2vec embeddings 
for the keyword column.
The model is first loaded, with various model parameters and trained on the input overview
followed by embedding creation. The size of the embedded vector is 768.
"""
def create_embeddings(movies_df):
  movies_df['overview'] = movies_df['overview'].astype(str)
  all_keywords = ' '.join(movies_df['overview']).split()
  model = Word2Vec(sentences=[all_keywords], vector_size=768, min_count=1, workers=3, epochs=5, window=5)
  word_embeddings = {}

  def get_embedding(keywords):
    keyword_list = keywords.split()
    embeddings = []
    for word in keyword_list:
      if word in word_embeddings:
        embeddings.append(word_embeddings[word])
      else:
        try:
          embedding = model.wv[word]
          word_embeddings[word] = embedding
          embeddings.append(embedding)
        except KeyError:
          pass
    if embeddings:
      return np.mean(embeddings, axis=0)
    else:
      return np.zeros(model.vector_size)

  movies_df['overview_embedding'] = movies_df['overview'].apply(get_embedding)

create_embeddings(movies_df)
```

```python
movies_df['overview_embedding'] = movies_df['overview_embedding'].apply(lambda x: str(x))
movies_df.loc[0, 'overview_embedding']
```

```python
# convert the embedded column array to embedded string and clean the overview_embedding_string column and remove any commas, whitespaces and unwanted characters
movies_df['overview_embedding_string'] = movies_df['overview_embedding'].str.replace('[', '').str.replace(']', '').str.replace('\n', '').str.replace(',', '')
```
