```python
"""
This file uses the data generated from running the iterations for each model for every epsilon value
and plot graphs for various result metrics.

All the result data was generated during the iteration step is processed in this code and used to plot graphs for result metrics.
"""

# import required libraries
import matplotlib.pyplot as plt
import numpy as np
```

```python
# create a list with values from 0 to 300 to plot iteration values on graphs
iterationList = list(range(1, 301))
```

```python
# gpt2

model = 'gpt2'

with open(f'/content/drive/MyDrive/LLM-NEW-2/{model}-embed/final-results/EP-0.01/{model}_plot_reward_data_EP-0.01.txt', 'r') as file:
    list_contents1 = [line.strip() for line in file]
with open(f'/content/drive/MyDrive/LLM-NEW-2/{model}-embed/final-results/EP-0.05/{model}_plot_reward_data_EP-0.05.txt', 'r') as file:
    list_contents2 = [line.strip() for line in file]
with open(f'/content/drive/MyDrive/LLM-NEW-2/{model}-embed/final-results/EP-0.1/{model}_plot_reward_data_EP-0.1.txt', 'r') as file:
    list_contents3 = [line.strip() for line in file]

list_contents1 = [ float(x) for x in list_contents1]
wor2vecRewardsList1 = list_contents1

list_contents2 = [ float(x) for x in list_contents2]
wor2vecRewardsList2 = list_contents2

list_contents3 = [ float(x) for x in list_contents3]
wor2vecRewardsList3 = list_contents3

plt.plot(iterationList, wor2vecRewardsList1, '-', label='ep=0.01')
plt.plot(iterationList, wor2vecRewardsList2, '-', label='ep=0.05')
plt.plot(iterationList, wor2vecRewardsList3, '-', label='ep=0.1')

line_value = 38000
plt.axhline(y=line_value, color='r', linestyle='-', label=f'Maximum rewards acheivable per iteration={line_value}')
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.ylim(0, 40000)

plt.xlabel('Number of Iterations', fontsize=20)
plt.ylabel('Cumulative Rewards', fontsize=20)
plt.title('Number of Iterations vs Cumulative Rewards Obtained\n GPT-2', fontsize=18)
plt.legend()
plt.show()
```

# Conversion Rate

```python
#  all modesl, epsilon = 0.01

dictEpZeroPointZeroOne = {}
dictEpZeroPointZerFive = {}
dictEpZeroPointOne = {}

model = 'word2vec'
with open(f'/content/drive/MyDrive/LLM-NEW-2/{model}-embed/final-results/EP-0.01/{model}_plot_conversion_data_EP-0.01.txt', 'r') as file:
    lines = file.readlines()
    list_contents1 = lines
# print(f'The conversion rate for {model} with epsilon value eqaul to 0.05 is:')
for line in list_contents1:
  val = line.split()
  dictEpZeroPointZeroOne[val[0]] = round(float(val[1]), 3)
  # print(f'Action {val[0]} Conversion rate {round(float(val[1]), 2)}')
sorted_dict1 = dict(sorted(dictEpZeroPointZeroOne.items(), key=lambda item: int(item[0])))
x_axis_1 = list(sorted_dict1.keys())
y_axis_1 = list(sorted_dict1.values())
# for key in lines:
#   print(key)

model = 'gpt2'
with open(f'/content/drive/MyDrive/LLM-NEW-2/{model}-embed/final-results/EP-0.01/{model}_plot_conversion_data_EP-0.01.txt', 'r') as file:
    lines = file.readlines()
    list_contents2 = lines
# print(f'The conversion rate for {model} with epsilon value eqaul to 0.05 is:')
for line in list_contents2:
  val = line.split()
  dictEpZeroPointZerFive[val[0]] = round(float(val[1]), 3)
  # print(f'Action {val[0]} Conversion rate {round(float(val[1]), 2)}')
sorted_dict2 = dict(sorted(dictEpZeroPointZerFive.items(), key=lambda item: int(item[0])))
x_axis_2 = list(sorted_dict2.keys())
y_axis_2 = list(sorted_dict2.values())
# for key in sorted_dict2.items():
#   print(key)

model = 'bert'
with open(f'/content/drive/MyDrive/LLM-NEW-2/{model}-embed/final-results/EP-0.01/{model}_plot_conversion_data_EP-0.01.txt', 'r') as file:
    lines = file.readlines()
    list_contents3 = lines
# print(f'The conversion rate for {model} with epsilon value eqaul to 0.05 is:')
for line in list_contents3:
  val = line.split()
  dictEpZeroPointOne[val[0]] = round(float(val[1]), 3)
  # print(f'Action {val[0]} Conversion rate {round(float(val[1]), 2)}')
sorted_dict3 = dict(sorted(dictEpZeroPointOne.items(), key=lambda item: int(item[0])))
x_axis_3 = list(sorted_dict3.keys())
y_axis_3 = list(sorted_dict3.values())
# for key in sorted_dict3.items():
#   print(key)


plt.figure(figsize=(10, 6))

# Plot data from the first dictionary (red line)
plt.plot(x_axis_1, y_axis_1, marker='o', color='red', label='Word2Vec', linestyle='solid')

# Plot data from the second dictionary (blue line)
plt.plot(x_axis_2, y_axis_2, marker='o', color='blue', label='GPT-2', linestyle='dashed')

# Plot data from the second dictionary (blue line)
plt.plot(x_axis_3, y_axis_3, marker='o', color='green', label='BERT', linestyle='dashdot')

# Customize the plot
plt.xlabel('Actions', fontsize=20)
plt.ylabel('Conversion Rate', fontsize=20)
plt.title('Conversion Rates for Different Actions For Epsilon = 0.01', fontsize=18)
plt.xticks(rotation=45, ha='right', fontsize=18)
plt.yticks(fontsize=18)
plt.grid(True)
plt.legend()  # Add legend to distinguish lines
plt.tight_layout()
plt.ylim(0, 0.8)

plt.show()
```
